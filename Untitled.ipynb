```python
from pathlib import Path
import pandas as pd

# These paths are relative to the repo's main folder.
data_dir = Path("data") / "processed"

master = pd.read_parquet(data_dir / "master.parquet")
episodes = pd.read_parquet(data_dir / "depeg_episodes.parquet")

print("Master dataset — rows, columns:", master.shape)
print("Episodes dataset — rows, columns:", episodes.shape)

display(master.head())
display(episodes.head())
```

```python
print("MASTER COLUMNS")
print(master.columns.tolist())

print("\nMASTER DATA TYPES")
print(master.dtypes)

print("\nEPISODE COLUMNS")
print(episodes.columns.tolist())
```

```python
import numpy as np
import pandas as pd
%pip install scikit-learn
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, brier_score_loss, roc_auc_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# 1. Feature sets
market_features = [
    "peg_deviation_bps",
    "log_return_1h",
    "log_return_6h",
    "log_return_24h",
    "realised_vol_6h",
    "realised_vol_24h",
    "volume_pct_change_1h",
    "volume_pct_change_6h",
    "volume_pct_change_24h",
]

onchain_features = [
    "tx_count_logdiff_1h",
    "tx_count_logdiff_6h",
    "tx_count_logdiff_24h",
    "tx_volume_logdiff_1h",
    "tx_volume_logdiff_6h",
    "tx_volume_logdiff_24h",
    "active_senders_logdiff_1h",
    "active_senders_logdiff_6h",
    "active_senders_logdiff_24h",
    "active_receivers_logdiff_1h",
    "active_receivers_logdiff_6h",
    "active_receivers_logdiff_24h",
]

combined_features = market_features + onchain_features

# 2. Target and eligibility filter (24h horizon)
LABEL = "label_h24"
ELIGIBLE = "prediction_eligible_market_onchain_h24"

# Filter to eligible sample where both market & on-chain data exist
df_sub = master[master[ELIGIBLE] == True].copy()
df_sub = df_sub.dropna(subset=combined_features + [LABEL, "hour"])

# 3. Time-based splits with 24-hour purge buffer
purge_gap = pd.Timedelta(hours=24)

train_mask = df_sub["hour"] < (pd.Timestamp("2023-01-01 00:00:00", tz="UTC") - purge_gap)
val_mask = (df_sub["hour"] >= pd.Timestamp("2023-01-01 00:00:00", tz="UTC")) & (
    df_sub["hour"] < (pd.Timestamp("2024-01-01 00:00:00", tz="UTC") - purge_gap)
)
test_mask = df_sub["hour"] >= pd.Timestamp("2024-01-01 00:00:00", tz="UTC")

train_df = df_sub[train_mask]
val_df = df_sub[val_mask]
test_df = df_sub[test_mask]

print(f"Sample sizes -> Train: {len(train_df)}, Val: {len(val_df)}, Test: {len(test_df)}")
print(f"Positive labels -> Train: {train_df[LABEL].sum()}, Val: {val_df[LABEL].sum()}, Test: {test_df[LABEL].sum()}")

# 4. Model fitting function
def evaluate_model(features, model_name):
    clf = make_pipeline(
        StandardScaler(),
        LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42),
    )
    clf.fit(train_df[features], train_df[LABEL])

    results = []
    for split_name, split_data in [("Validation (2023)", val_df), ("Test (2024-2025)", test_df)]:
        probs = clf.predict_proba(split_data[features])[:, 1]
        y_true = split_data[LABEL]

        roc = roc_auc_score(y_true, probs)
        pr_auc = average_precision_score(y_true, probs)
        brier = brier_score_loss(y_true, probs)

        results.append({
            "Model": model_name,
            "Split": split_name,
            "ROC-AUC": round(roc, 4),
            "PR-AUC": round(pr_auc, 4),
            "Brier Score": round(brier, 4),
        })
    return clf, pd.DataFrame(results)

# Train and evaluate both models on identical coin-hours
model_mkt, res_mkt = evaluate_model(market_features, "Market Only")
model_all, res_all = evaluate_model(combined_features, "Market + On-Chain")

summary_table = pd.concat([res_mkt, res_all], ignore_index=True)
display(summary_table)
```
